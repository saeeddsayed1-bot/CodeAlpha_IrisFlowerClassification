Iris Flower Classification

Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import sklearn
from sklearn.preprocessing import StandardScaler,OneHotEncoder,LabelEncoder
from sklearn.linear_model import LinearRegression , LogisticRegression
from sklearn.tree import DecisionTreeClassifier , DecisionTreeRegressor
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.cluster import KMeans
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split , cross_val_score
from sklearn.datasets import make_moons
from sklearn.metrics import confusion_matrix, classification_report, precision_score, recall_score, f1_score , accuracy_score, silhouette_score, mean_absolute_error, mean_squared_error, r2_score
```

Load Dataset

```python
df = pd.read_csv(r"C:\Users\Saee\Downloads\Iris.xls")
```

Explore Dataset

```python
print(df.head())
print(df.info())
print(df.describe())
print(df.isnull().sum())
print(df['Species'].unique())
print(df['Species'].value_counts())
print(df.columns)
```

Data Visualization

```python
plt.figure(figsize=(14,10))
plt.subplot(2,2,1)
plt.subplots_adjust(hspace=0.5,wspace=0.5)
sns.barplot(x='Species', y='SepalLengthCm', data=df)
plt.subplot(2,2,2)
sns.barplot(x='Species', y='SepalWidthCm', data=df)
plt.subplot(2,2,3)
sns.barplot(x='Species', y='PetalLengthCm', data=df)
plt.subplot(2,2,4)
sns.barplot(x='Species', y='PetalWidthCm', data=df)
```

Data Preprocessing

```python
df_new = df.drop('Id', axis=1)
df_new.head()
```

Label Encoding

```python
le = LabelEncoder()
df_new['Species'] = le.fit_transform(df_new['Species'])
df_new.head()
```

Define Features and Target

```python
x = df_new.drop("Species" , axis= 1)
y = df_new["Species"]
```

Train Test Split

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

Model Building (Logistic Regression)

```python
LR = LogisticRegression()
```

Model Training & Prediction

```python
LR.fit(x_train , y_train)
y_pred = LR.predict(x_test)
```

Model Evaluation

```python
accuracy = accuracy_score(y_test , y_pred)
print("Accuracy:", accuracy)
```

Classification Report

```python
CR = classification_report(y_test , y_pred)
print(CR)
```

Confusion Matrix

```python
CM = confusion_matrix(y_test , y_pred)
print(CM)
```

Conclusion

The Iris Flower Classification project was successfully completed using Logistic Regression. The model achieved an accuracy of 100%, indicating excellent performance in classifying flower species. Data visualization showed that Petal Length and Petal Width were the most important features for distinguishing between species. The confusion matrix and classification report confirmed that the model made accurate predictions with no classification errors. This project demonstrated the complete machine learning workflow, from data preprocessing to model evaluation.
